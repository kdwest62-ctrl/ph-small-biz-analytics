```python
import pandas as pd

csv_name = input('CSV name: ')
df = pd.read_csv(csv_name)
print(df)
choice = input('Use (f) full or (s) select date range: ')
if choice == 's':
  start_index = int(input('Start date (index): '))
  end_index = int(input('End date (index): '))
  end_index += 1
  df = df.iloc[start_index:end_index].copy()
```

```python
df['revenue'] = df['quantity_sold'] * df['price']
total_revenue = df.groupby('product')['revenue'].sum().reset_index()
sorted_total_revenue = total_revenue.sort_values(by='revenue', ascending=False)
print(sorted_total_revenue)
```

```python
total_sold = df.groupby('product')['quantity_sold'].sum().reset_index()
sorted_total_sold = total_sold.sort_values(by='quantity_sold', ascending=False)
print(sorted_total_sold)
```

```python
total_sold = df.groupby('product')['quantity_sold'].sum().reset_index()
sorted_total_sold = total_sold.sort_values(by='quantity_sold')
print(sorted_total_sold)
```
