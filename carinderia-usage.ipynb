Problem: Carinderias often overbuy vegetables/meat, leading to spoilage.

Why useful: reduces food waste and saves money.

```python
# decide whether to use full CSV or select preferred date range

import pandas as pd

csv_name = input('CSV name: ')
df = pd.read_csv(csv_name)
print(df)
choice = input('Use (f) full or (s) select date range: ')
if choice == 's':
  start_index = int(input('Start date (index): '))
  end_index = int(input('End date (index): '))
  end_index += 1
  df = df.iloc[start_index:end_index].copy()
```

```python
# forecast next day’s ingredient needs with simple moving average

import numpy as np

def moving_average(d, window_size):
  weights = np.ones(window_size) / window_size
  return np.convolve(d, weights, mode='valid')

ingredient = input('Ingredient: ')
data = np.array(df[ingredient].tolist())
print(f'Daily usage: {data}')
size = int(input('Window size: '))
print(f'Moving average: {moving_average(data, size)}')
```
